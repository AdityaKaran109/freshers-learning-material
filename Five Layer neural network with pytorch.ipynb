```python
import torch
import torch.nn as nn
```

```python
class FiveLayerNetwork(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(FiveLayerNetwork, self).__init__()

        self.layer1 = nn.Linear(input_size, hidden_size)
        self.layer2 = nn.Linear(hidden_size, hidden_size)
        self.layer3 = nn.Linear(hidden_size, hidden_size)
        self.layer4 = nn.Linear(hidden_size, hidden_size)
        self.layer5 = nn.Linear(hidden_size, output_size)
        
        self.relu = nn.ReLU()

    def forward(self, x):
            x = self.layer1(x)
            x = self.relu(x)
            x = self.layer2(x)
            x = self.relu(x)
            x = self.layer3(x)
            x = self.relu(x)
            x = self.layer4(x)
            x = self.relu(x)
            out = self.layer5(x)
            
            return out
```

```python

INPUT_FEATURES = 10     
NEURONS_PER_LAYER = 64  
OUTPUT_PREDICTIONS = 1 

model = FiveLayerNetwork(
    input_size=INPUT_FEATURES, 
    hidden_size=NEURONS_PER_LAYER, 
    output_size=OUTPUT_PREDICTIONS
)

print(model)

dummy_data = torch.randn(5, INPUT_FEATURES)
predictions = model(dummy_data)
print("\nRaw Output Predictions:")
print(predictions)
```

nn.Linear (The Connections): Think of these as the wiring between the nodes. When data passes through a linear layer, the network multiplies the input by "weights" and adds a "bias". During training, the network tweaks these weights to get smarter.

nn.ReLU (The Activation): If we only used linear layers, our 5-layer network would mathematically collapse into a 1-layer network. The ReLU function (which simply turns all negative numbers to 0) introduces "non-linearity," which allows the network to learn complex, curved patterns instead of just straight lines.

The forward pass: This is the chronological sequence of events. You are literally telling Python: Take x, push it through layer 1, apply ReLU, update x, push it to layer 2...
